```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
# My favorite movies
movies = {
    'Title': ['The Matrix', 'Inception', 'Interstellar', 'The Dark Knight', 'Parasite'],
    'Year': [1999, 2010, 2014, 2008, 2019],
    'Rating': [9, 8.5, 9, 9.5, 8],
    'Genre': ['Sci-Fi', 'Sci-Fi', 'Sci-Fi', 'Action', 'Thriller']
}

# Create a DataFrame
df = pd.DataFrame(movies)
df
```

```python
# Basic statistics
print(f"Average rating: {df['Rating'].mean():.1f}")
print(f"Highest rated: {df.loc[df['Rating'].idxmax(), 'Title']}")
print(f"Lowest rated: {df.loc[df['Rating'].idxmin(), 'Title']}")
```

```python
# Create a bar chart
plt.figure(figsize=(10, 5))
plt.bar(df['Title'], df['Rating'], color='skyblue')
plt.xlabel('Movie Title')
plt.ylabel('My Rating')
plt.title('My Favorite Movies Ranked')
plt.xticks(rotation=45, ha='right')
plt.ylim(0, 10)
plt.tight_layout()
plt.show()
```
